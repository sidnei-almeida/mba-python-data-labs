## 6. Pergunta 1 — Quais itens aparecem mais vezes?

Para contar categorias, usamos `.value_counts()`.

Essa pergunta não mede faturamento.  
Ela mede **quantidade de registros/vendas** por item.

````python
# Contagem de vendas por item.

vendas_por_item = df["Item"].value_counts()

vendas_por_item
````

### Primeiro gráfico: barras com Pandas

O Pandas consegue criar gráficos simples usando `.plot()`.

Por baixo, ele usa o Matplotlib.

````python
# Criando um gráfico de barras com Pandas.

vendas_por_item.plot(kind="bar", figsize=(10, 5))

plt.title("Quantidade de Vendas por Item")
plt.xlabel("Item")
plt.ylabel("Quantidade de Vendas")
plt.xticks(rotation=45)
plt.show()
````

## 7. Entendendo a estrutura básica do Matplotlib

Um gráfico simples costuma ter esta estrutura:

```python
plt.figure(figsize=(largura, altura))
plt.bar(x, y)
plt.title("Título")
plt.xlabel("Nome do eixo X")
plt.ylabel("Nome do eixo Y")
plt.show()
```

Pense assim:

- `plt.figure()` cria a área do gráfico;
- `plt.bar()` cria barras;
- `plt.title()` coloca título;
- `plt.xlabel()` nomeia o eixo X;
- `plt.ylabel()` nomeia o eixo Y;
- `plt.show()` exibe o gráfico.

````python
# Criando o mesmo gráfico, agora usando Matplotlib de forma mais explícita.

plt.figure(figsize=(10, 5))

plt.bar(vendas_por_item.index, vendas_por_item.values)

plt.title("Quantidade de Vendas por Item")
plt.xlabel("Item")
plt.ylabel("Quantidade de Vendas")
plt.xticks(rotation=45)

plt.show()
````

## 8. Pergunta 2 — Qual item gerou mais faturamento?

Agora queremos somar o faturamento por item.

Para isso usamos `.groupby()`.

A lógica é:

```python
df.groupby("Item")["Calculated Total"].sum()
```

Tradução:

> agrupe por item e some o valor calculado.

````python
# Faturamento por item.

faturamento_por_item = df.groupby("Item")["Calculated Total"].sum()

# Ordenando do maior para o menor.
faturamento_por_item = faturamento_por_item.sort_values(ascending=False)

faturamento_por_item
````

````python
# Gráfico de faturamento por item.

plt.figure(figsize=(10, 5))

plt.bar(faturamento_por_item.index, faturamento_por_item.values)

plt.title("Faturamento por Item")
plt.xlabel("Item")
plt.ylabel("Faturamento")
plt.xticks(rotation=45)

plt.show()
````

## 9. Melhorando a leitura dos números

Às vezes, o resultado aparece com muitas casas decimais.

Podemos usar `.round(2)` para arredondar.

Também podemos transformar o resultado em DataFrame com `.reset_index()`.

````python
# Transformando a Series em DataFrame.

faturamento_por_item_df = faturamento_por_item.reset_index()

# Renomeando as colunas.
faturamento_por_item_df.columns = ["Item", "Faturamento"]

# Arredondando o faturamento.
faturamento_por_item_df["Faturamento"] = faturamento_por_item_df["Faturamento"].round(2)

faturamento_por_item_df
````

## 10. Pergunta 3 — Qual é o ticket médio por item?

O **ticket médio** é o valor médio gasto.

Aqui vamos calcular a média de `Calculated Total` por item:

```python
df.groupby("Item")["Calculated Total"].mean()
```

````python
# Ticket médio por item.

ticket_medio_por_item = df.groupby("Item")["Calculated Total"].mean()

ticket_medio_por_item = ticket_medio_por_item.sort_values(ascending=False).round(2)

ticket_medio_por_item
````

````python
# Gráfico de ticket médio por item.

plt.figure(figsize=(10, 5))

plt.bar(ticket_medio_por_item.index, ticket_medio_por_item.values)

plt.title("Ticket Médio por Item")
plt.xlabel("Item")
plt.ylabel("Ticket Médio")
plt.xticks(rotation=45)

plt.show()
````

## 11. Pergunta 4 — Qual método de pagamento é mais usado?

## 12. Gráfico de pizza

O gráfico de pizza pode ser usado para mostrar participação percentual.

## 14. Pergunta 6 — Como o faturamento varia por mês?

Como temos as colunas `Year` e `Month`, podemos analisar o faturamento ao longo do tempo.

Vamos agrupar por mês e somar `Calculated Total`.

````python
# Faturamento por mês.

faturamento_por_mes = df.groupby("Month")["Calculated Total"].sum()

faturamento_por_mes
````

### Gráfico de linha

Gráficos de linha são úteis para mostrar evolução ao longo do tempo.

A estrutura básica é:

```python
plt.plot(x, y, marker="o")
```

O `marker="o"` coloca um ponto em cada mês.

````python
# Gráfico de linha do faturamento por mês.

plt.figure(figsize=(10, 5))

plt.plot(faturamento_por_mes.index, faturamento_por_mes.values, marker="o")

plt.title("Faturamento por Mês")
plt.xlabel("Mês")
plt.ylabel("Faturamento")
plt.xticks(faturamento_por_mes.index)

plt.show()
````

## 15. Pergunta 7 — Quantidade vendida por mês

## 16. Comparando duas métricas com `.agg()`

O método `.agg()` permite calcular várias métricas de uma vez.

Exemplo:

```python
df.groupby("Item").agg({
    "Quantity": "sum",
    "Calculated Total": "sum"
})
```

Isso agrupa por item e calcula:

- soma da quantidade;
- soma do faturamento.

````python
# Agregando quantidade e faturamento por item.

resumo_por_item = df.groupby("Item").agg({
    "Quantity": "sum",
    "Calculated Total": "sum"
})

# Ordenando pelo faturamento.
resumo_por_item = resumo_por_item.sort_values("Calculated Total", ascending=False)

resumo_por_item
````

## 17. Renomeando colunas do resumo

Para deixar a tabela mais clara, podemos renomear as colunas.

````python
# Criando uma cópia do resumo para apresentação.

resumo_por_item_formatado = resumo_por_item.copy()

# Renomeando colunas.
resumo_por_item_formatado = resumo_por_item_formatado.rename(columns={
    "Quantity": "Quantidade Vendida",
    "Calculated Total": "Faturamento"
})

# Arredondando faturamento.
resumo_por_item_formatado["Faturamento"] = resumo_por_item_formatado["Faturamento"].round(2)

resumo_por_item_formatado
````
